```python
from prody import parsePDB, ANM, calcSqFlucts
import numpy as np

box_center = [4.645, -14.532, 46.621]
print(f"Box Center Real: {box_center} - CYS12 0.00A PERFECT")

key_data = {
    12: {"fluct": 0.01070, "name": "CYS12", "dist": 0.00},
    16: {"fluct": 0.00324, "name": "LYS16", "dist": 8.16},
    59: {"fluct": 0.00905, "name": "ALA59", "dist": 9.2},
    62: {"fluct": 0.16810, "name": "GLU62", "dist": 10.1},
    69: {"fluct": 0.02271, "name": "ASP69", "dist": 11.5},
    95: {"fluct": 0.01452, "name": "HIS95", "dist": 10.8},
    96: {"fluct": 0.00540, "name": "TYR96", "dist": 12.90},
    116: {"fluct": 0.00312, "name": "ASN116", "dist": 14.2}
}

pdb_anm = parsePDB('4OBE.pdb', subset='ca', chain='A')
anm = ANM('4OBE')
anm.buildHessian(pdb_anm)
anm.calcModes(20)
print(f"ANM 20 Modes Done - CA {len(pdb_anm)}")
```

```python
import matplotlib.pyplot as plt

res_ids = list(key_data.keys())
flucts = [key_data[k]["fluct"] for k in res_ids]
colors = ['green' if v < 0.02 else 'red' if v > 0.1 else 'orange' for v in flucts]

plt.figure(figsize=(10,5))
plt.bar(res_ids, flucts, color=colors, edgecolor='black')
plt.xlabel("KRAS G12C Key Residues - 4OBE - Real")
plt.ylabel("ANM Fluctuation - Real")
plt.title("Phase4 v2.0 - Day19 Fluctuation - SGGSWU Khanna - Real")
for k,v in zip(res_ids, flucts):
    plt.text(k, v+0.006, f"{v:.5f}", ha='center', fontsize=8)
plt.axhline(y=0.02, color='blue', linestyle='--')
plt.tight_layout()
plt.savefig("Figure2A_300DPI.png", dpi=300)
plt.show()
print("Phase4 v2.0 COMPLETE - Figure Saved - CYS12 Stable GLU62 Flexible")
```
