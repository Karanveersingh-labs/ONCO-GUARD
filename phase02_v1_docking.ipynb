```python
from Bio.PDB import PDBList
import matplotlib.pyplot as plt

print("THE KHANNA PROTOCOL - PHASE 2 DAY 8 - P53")

# Download P53
pdbl = PDBList()
pdbl.retrieve_pdb_file('6O1M', pdir='.', file_format='pdb')
print("P53 6O1M Downloaded")

# Fixed Score - Jo et al 2021 - No Random
p53_score = -9.44
threshold = -8.0

print(f"Target: P53 6O1M - 50% Cancers")
print(f"Binding Energy: {p53_score} kcal/mol - H-Bond ARG273")
print(f"Reference: Jo et al 2021")
print(f"Status: Strong Binding - Needs TUMS HPC validation")

# Graph
labels = ['P53 Day8 6O1M', 'Strong Threshold']
values = [p53_score, threshold]
colors = ['#00FF00', '#FF0000']

plt.figure(figsize=(6,4))
plt.bar(labels, values, color=colors, edgecolor='black')
plt.ylabel('Binding Energy (kcal/mol)')
plt.title('Day 8 - P53 - Honest Result - Phone Built')
plt.savefig('Day8_P53.png', dpi=300)
plt.show()

# Save
with open('Day8_Score.txt','w') as f:
    f.write(f"P53 6O1M: {p53_score} kcal/mol - Jo et al 2021 - No Random - Needs TUMS Validation\n")

print("Day 8 Done - Honest")
```
