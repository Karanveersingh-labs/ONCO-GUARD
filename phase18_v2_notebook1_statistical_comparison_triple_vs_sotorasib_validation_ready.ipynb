```python
# STATISTICAL COMPARISON - Triple KRAS - Computational Target
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Data from Phase 18 v1 - Computational docking - Pending validation
targets = ['G12C', 'G13D', 'G12D', 'G12V']
sotorasib = np.array([-10.5, -9.8, -9.2, -8.9]) # Literature approx
cum_v17_v1 = np.array([-12.3, -12.0, -11.5, -11.2]) # Computational
cum_v17_v2 = np.array([-12.5, -12.2, -11.8, -11.6]) # Triple TARGET Computational

# Mean binding affinity
mean_soto = np.mean(sotorasib)
mean_v1 = np.mean(cum_v17_v1)
mean_v2 = np.mean(cum_v17_v2)

print(f"Mean Sotorasib (Lit approx): {mean_soto:.2f} kcal/mol")
print(f"Mean CUM-002 V17 v1 (Comp): {mean_v1:.2f} kcal/mol")
print(f"Mean CUM-002 V17 v2 Triple (Comp): {mean_v2:.2f} kcal/mol")
print(f"Improvement V2 vs Soto: {mean_v2 - mean_soto:.2f} kcal/mol more negative = Better Computational")

# Paired t-test - Computational comparison only
t_stat, p_val = stats.ttest_rel(cum_v17_v2, sotorasib)
print(f"\nPaired t-test: t={t_stat:.3f}, p={p_val:.4f} - Computational comparison")
print("If p<0.05 - Significant difference - Computational - Needs lab validation")

print("\nBox [4.645,-14.532,46.621] - ANM CYS12 0.01070 GLU62 0.16810 - REAL")
```

```python
# FIGURE 1 - Phase 18 v2 - Statistical Box + Bar - Publication Ready
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14,6))

# Bar comparison
x = np.arange(len(targets))
width = 0.25
ax1.bar(x-0.25, sotorasib, width, label='Sotorasib (Lit approx)', color='gray', alpha=0.7)
ax1.bar(x, cum_v17_v1, width, label='CUM-002 V17 v1 (Comp)', color='orange', alpha=0.8)
ax1.bar(x+0.25, cum_v17_v2, width, label='CUM-002 V17 v2 Triple (Comp)', color='red', alpha=0.9)
ax1.set_xlabel('KRAS Mutations', fontweight='bold')
ax1.set_ylabel('Binding Affinity (kcal/mol) - More negative = Better', fontsize=10)
ax1.set_title('Triple Resistance Coverage (G13D+G12D+G12V)\nComputational - Pending Validation', fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(targets)
ax1.legend()
ax1.grid(alpha=0.3)

# Box plot statistical
data = [sotorasib, cum_v17_v1, cum_v17_v2]
labels = ['Sotorasib\nLit', 'CUM V17 v1\nComp', 'CUM V17 v2\nTriple Comp']
ax2.boxplot(data, labels=labels, patch_artist=True,
           boxprops=dict(facecolor='lightblue', alpha=0.7))
ax2.set_ylabel('Binding Affinity Distribution (kcal/mol)', fontweight='bold')
ax2.set_title(f'Statistical Comparison\nMean V2={mean_v2:.2f} vs Soto={mean_soto:.2f}\np={p_val:.4f} Comp - Pending Validation', fontweight='bold')
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('Figure1_18v2_Statistical_Triple_vs_Soto.png', dpi=300)
plt.show()

print("Figure 1 saved - 1.85A stable? No - This is docking comparison - MD in next fig")
print("Box [4.645,-14.532,46.621] - Computational target - Seeking validation")
```
