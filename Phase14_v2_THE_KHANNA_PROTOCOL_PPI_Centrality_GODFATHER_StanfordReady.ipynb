```python
import pandas as pd, matplotlib.pyplot as plt, networkx as nx
plt.close('all')
print("Phase14 v2 GODFATHER Started")
edges = [["KRAS","BRAF",0.999],["KRAS","RAF1",0.998],["KRAS","MAPK1",0.95],["KRAS","AKT1",0.92],["BRAF","MAP2K1",0.995],["RAF1","MAP2K1",0.99],["MAP2K1","MAPK1",0.999],["PIK3CA","AKT1",0.998]]
df = pd.DataFrame(edges, columns=['A','B','Score'])
G = nx.Graph()
for a,b,s in edges:
  G.add_edge(a,b,weight=s)
  deg = nx.degree_centrality(G)
  bet = nx.betweenness_centrality(G)
  res = pd.DataFrame([{"Protein":k,"Degree":deg[k],"Betweenness":bet[k]} for k in G.nodes()])
  res = res.sort_values('Degree', ascending=False)
  res.to_csv('Phase14_v2_Centrality_GODFATHER.csv', index=False)
  print(res)
  plt.figure(figsize=(10,7))
  pos = nx.spring_layout(G, seed=42)
  nx.draw_networkx_nodes(G, pos, node_size=[v*3000+500 for v in deg.values()], node_color='lightcoral', edgecolors='black')
  nx.draw_networkx_edges(G, pos, width=2, edge_color='gray')
  nx.draw_networkx_labels(G, pos, font_weight='bold')
  plt.title('Phase14 v2 GODFATHER - PPI Network - KRAS Master Hub')
  plt.axis('off')
  plt.tight_layout()
  plt.savefig('Phase14_v2_PPI_Network_GODFATHER.png', dpi=300)
  plt.show()
  print("Phase14 v2 DONE")
```
