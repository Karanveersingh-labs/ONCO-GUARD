```python
import pandas as pd
import matplotlib.pyplot as plt
binding_pocket = {'CYS12': 12, 'ARG68': 68, 'ASP92': 92}
amino = {'CYS': 86.0, 'ASP': 111.0, 'VAL': 140.0, 'ARG': 173.0, 'GLY': 60.1}
mutations = [('CYS12','ASP12','CYS','ASP'),('CYS12','VAL12','CYS','VAL'),('ARG68','GLY68','ARG','GLY')]
results = [[f"{o}->{m}", amino[ma]-amino[oa], ((146.5+(amino[ma]-amino[oa])*0.25-146.5)/146.5)*100] for o,m,oa,ma in mutations]
for r in results:
 print(f"{r[0]}: Change {r[2]:.2f}%")
 df = pd.DataFrame(results, columns=['mutation','delta_vol','change_percent'])
 df.to_csv('Phase13_v1_Results.csv', index=False)
 print(df)
 plt.bar(df['mutation'], df['change_percent'])
 plt.ylabel('Pocket Change %')
 plt.title('Phase13 v1 - KRAS G12C - Khanna Protocol')
 plt.savefig('Phase13_v1_Pocket_Impact.png', dpi=200)
 plt.show()
 print("Phase13 v1 DONE")
```
