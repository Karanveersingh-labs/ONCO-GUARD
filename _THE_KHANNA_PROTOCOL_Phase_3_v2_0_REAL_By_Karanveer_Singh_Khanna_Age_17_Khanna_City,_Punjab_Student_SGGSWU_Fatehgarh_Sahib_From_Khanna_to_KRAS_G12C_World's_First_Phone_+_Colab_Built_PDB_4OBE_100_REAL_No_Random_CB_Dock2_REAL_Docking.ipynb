```python
from Bio.PDB import PDBIO, Select

class ChainA_Select(Select):
    def accept_chain(self, chain):
        return chain.id == 'A'
    def accept_residue(self, residue):
        return residue.resname != 'HOH'

io = PDBIO()
io.set_structure(structure)
io.save("KRAS_Day13_v2.pdb", ChainA_Select())
print("KRAS_Day13_v2.pdb Saved")
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem, Descriptors
smiles = "CC(C)C1=CC=C(C=C1)C=O"
mol = Chem.MolFromSmiles(smiles)
mol = Chem.AddHs(mol)
AllChem.EmbedMolecule(mol, AllChem.ETKDGv3())
AllChem.UFFOptimizeMolecule(mol)
writer = Chem.SDWriter('Cuminaldehyde_Day13_v2.sdf')
writer.write(mol)
writer.close()
Chem.MolToPDBFile(mol, 'Cuminaldehyde_Day13_v2.pdb')
print(f"MW {Descriptors.MolWt(mol):.2f} SDF+PDB Ready")
```
