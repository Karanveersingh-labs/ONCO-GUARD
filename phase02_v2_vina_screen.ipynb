```python
# FINAL FIX - No Meeko - Direct OpenBabel - 100% Works in Colab 2026

!apt-get install -y -qq openbabel
!pip install rdkit -q
!wget -q https://files.rcsb.org/download/4OBE.pdb -O 4OBE.pdb

from rdkit import Chem
from rdkit.Chem import AllChem

smiles = 'CC(C)c1ccc(C=O)cc1'
print(f"Cuminaldehyde SMILES: {smiles} - MW 148.20 REAL - Matches Phase 6")

mol = Chem.MolFromSmiles(smiles)
mol = Chem.AddHs(mol)
AllChem.EmbedMolecule(mol, randomSeed=42)
AllChem.MMFFOptimizeMolecule(mol)

Chem.MolToPDBFile(mol, 'ligand.pdb')
print("ligand.pdb READY")

# PDB to PDBQT - OpenBabel se - No meeko, No gemmi
!obabel ligand.pdb -O ligand.pdbqt -h --partialcharge gasteiger
!obabel 4OBE.pdb -O 4OBE.pdbqt -xr -h

print("Both PDBQT READY - REAL")
!ls -lh *.pdbqt
```

```python
import os
if os.path.exists("config.txt"):
    os.remove("config.txt")
if os.path.exists("REAL_LOG.txt"):
    os.remove("REAL_LOG.txt")
if os.path.exists("REAL_KRAS_4OBE.pdbqt"):
    os.remove("REAL_KRAS_4OBE.pdbqt")

with open('config.txt','w') as f:
    f.write("""receptor = 4OBE.pdbqt
ligand = ligand.pdbqt
center_x = 4.645
center_y = -14.532
center_z = 46.621
size_x = 20
size_y = 20
size_z = 20
out = REAL_KRAS_4OBE.pdbqt
exhaustiveness = 32
num_modes = 9
""")
print("config.txt READY - No log line - 100% FIXED")
!cat config.txt
```

```python
!vina --receptor 4OBE.pdbqt --ligand ligand.pdbqt --center_x 4.645 --center_y -14.532 --center_z 46.621 --size_x 20 --size_y 20 --size_z 20 --out REAL_KRAS_4OBE.pdbqt --exhaustiveness 32 --num_modes 9 > REAL_LOG.txt
!cat REAL_LOG.txt
!ls -lh 4OBE.pdbqt ligand.pdbqt REAL_KRAS_4OBE.pdbqt REAL_LOG.txt
!echo "===== REAL TABLE ====="
!grep -A 20 "mode.*affinity" REAL_LOG.txt
```
