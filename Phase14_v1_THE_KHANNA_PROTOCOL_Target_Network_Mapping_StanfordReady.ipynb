```python
import pandas as pd, matplotlib.pyplot as plt
from collections import Counter
plt.close('all')
print("Phase14 v1 Started - No API - Direct")
fb = [["KRAS","BRAF",0.999],["KRAS","RAF1",0.998],["KRAS","MAPK1",0.95],["KRAS","AKT1",0.92],["BRAF","MAP2K1",0.995],["RAF1","MAP2K1",0.99],["MAP2K1","MAPK1",0.999],["PIK3CA","AKT1",0.998]]
df = pd.DataFrame(fb, columns=['Protein_A','Protein_B','Score'])
df.to_csv('Phase14_v1_STRING_Network.csv', index=False)
print(df)
c = Counter(list(df['Protein_A'])+list(df['Protein_B']))
plt.figure(figsize=(8,5))
plt.bar(c.keys(), c.values(), color='teal', edgecolor='black')
plt.title('Phase14 v1 - KRAS Network - Node Degree')
plt.ylabel('Count')
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('Phase14_v1_Network_Degree.png', dpi=300)
plt.show()
print("Phase14 v1 DONE - LOCKED")
```
