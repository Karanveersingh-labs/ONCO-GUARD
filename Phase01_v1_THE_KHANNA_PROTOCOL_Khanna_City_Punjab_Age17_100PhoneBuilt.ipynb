```python
# Day 4 - FINAL - 100% WORKING - PhD Level Locked

protein = "MELACRVWGSGTQVDK" # 16aa
gc_table = {'M':'AUG','E':'GAG','L':'CUG','A':'GCG','C':'UGC','R':'CGC','V':'GUG','W':'UGG','G':'GGC','S':'AGC','T':'ACC','Q':'CAG','D':'GAC','K':'AAG'}

# Complete table - Sab codons - Koi missing nahi
codon_table = {
    'UUU':'F','UUC':'F','UUA':'L','UUG':'L','CUU':'L','CUC':'L','CUA':'L','CUG':'L',
    'AUU':'I','AUC':'I','AUA':'I','AUG':'M','GUU':'V','GUC':'V','GUA':'V','GUG':'V',
    'UCU':'S','UCC':'S','UCA':'S','UCG':'S','AGU':'S','AGC':'S',
    'CCU':'P','CCC':'P','CCA':'P','CCG':'P','ACU':'T','ACC':'T','ACA':'T','ACG':'T',
    'GCU':'A','GCC':'A','GCA':'A','GCG':'A','UAU':'Y','UAC':'Y','UAA':'*','UAG':'*','UGA':'*',
    'CAU':'H','CAC':'H','CAA':'Q','CAG':'Q','AAU':'N','AAC':'N','AAA':'K','AAG':'K',
    'GAU':'D','GAC':'D','GAA':'E','GAG':'E','UGU':'C','UGC':'C','UGG':'W',
    'CGU':'R','CGC':'R','CGA':'R','CGG':'R','AGA':'R','AGG':'R',
    'GGU':'G','GGC':'G','GGA':'G','GGG':'G'
}

def translate(rna):
    p=""
    for i in range(0,len(rna)-2,3):
        c=rna[i:i+3]
        aa=codon_table.get(c,'X')
        if aa=='*': break
        p+=aa
    return p

mrna_day2 = "".join(gc_table[aa] for aa in protein) + "UAA"
five_utr = "ACAUUUGCUCUCUGACACAACUGUGUUUACUGCAAC"
kozak = "GCCACC"
three_utr = "UGAUAUACUGACAAAAUUAACAGAAUAAAAGAA"
poly_a = "A"*30

full_day3 = five_utr + kozak + mrna_day2
full_vaccine = full_day3 + three_utr + poly_a

print(f"Day 2 CDS: {len(mrna_day2)}nt - GC: {(mrna_day2.count('G')+mrna_day2.count('C'))/len(mrna_day2)*100:.1f}%")
print(f"Day 3: {len(full_day3)}nt")
print(f"Day 4 FULL: {len(full_vaccine)}nt")
print(f"Target: {protein}")
print(f"Translated: {translate(mrna_day2)}")
print(f"MATCH: {translate(mrna_day2)==protein} ✅")

import matplotlib.pyplot as plt
days=['Day1\n45nt','Day2\n51nt','Day3\n93nt','Day4\n155nt']
lens=[45, len(mrna_day2), len(full_day3), len(full_vaccine)]
cols=['#ff6b6b','#feca57','#48dbfb','#1dd1a1']
plt.figure(figsize=(8,4.5))
bars=plt.bar(days,lens,color=cols,edgecolor='black',linewidth=1.2)
plt.ylabel('Length (nt)'); plt.title('ONCO-GUARD™ - Day 1 to Day 4',fontweight='bold')
for b,l in zip(bars,lens): plt.text(b.get_x()+b.get_width()/2,b.get_height()+2,f'{l}nt',ha='center',fontweight='bold')
plt.ylim(0,max(lens)+20); plt.tight_layout(); plt.show()
```
